Important : ce notebook ne modifie pas `memoirev3`. Il enrichit seulement les sorties McGill deja produites par la pipeline initiale.

```python
from pathlib import Path
import runpy

cwd = Path.cwd().resolve()
candidates = [
    cwd / "run_objective1_reinforcement.py",
    cwd.parent / "run_objective1_reinforcement.py",
    Path("/Users/alioubarry/PROJECT/mcgill_iot_cattle/run_objective1_reinforcement.py"),
]
script = next((p for p in candidates if p.exists()), None)
if script is None:
    raise FileNotFoundError("run_objective1_reinforcement.py introuvable")

project = script.parent
outdir = project / "reports" / "objective1_pipeline_icetag" / "renforcement_scientifique"
print("Script:", script)
print("Sorties:", outdir)
```
